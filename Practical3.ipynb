```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("heart.csv")
```

```python
data
```

```python
# Count values
target_counts = data['target'].value_counts()

# Plot
plt.figure(figsize=(6,6))
plt.pie(target_counts, labels=target_counts.index, autopct='%1.1f%%')
plt.title('Heart Disease Distribution')
plt.show()
```

```python
counts = data["cp"].value_counts()
plt.figure(figsize=(6,6))
plt.bar(counts.index, counts.values)
plt.title("cp count distribution")
plt.xlabel("cp")
plt.ylabel("count")
plt.show()
```

```python
counts = data["cp"].value_counts()
plt.figure(figsize=(6,6))
sns.barplot(x=counts.index, y=counts.values)
plt.title("cp count distribution")
plt.xlabel("cp")
plt.ylabel("count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(data['age'])
plt.title('Age Distribution')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='target', y='age', data=data)
plt.title('Age vs Target')
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(data['age'], bins=10)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(x='age', data=data, bins=10, kde=True)
plt.title('Age Distribution with KDE')
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.plot(data['age'], label='Age')
plt.plot(data['trestbps'], label='Resting BP')
plt.legend()
plt.title('Line Graph of Age & BP')
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.lineplot(x=data.index, y=data['age'], label='age')
plt.legend()
plt.title('Line Graph of Age & BP')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(data['age'], data['thalach'], alpha=0.2, s=20)
plt.title("scatter plot of age vs heart rate")
plt.xlabel("age")
plt.ylabel("heart rate")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='age', y='thalach', hue='target', data=data)
plt.title("scatter plot of age vs heart rate")
plt.xlabel("age")
plt.ylabel("heart rate")
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.fill_between(data.index, data['age'], alpha=0.3)

plt.title('Area Chart of Age')
plt.xlabel('Index')
plt.ylabel('Age')
plt.show()
```

```python
plt.figure(figsize=(8,4))

sns.lineplot(x=data.index, y=data['age'])
plt.fill_between(data.index, data['age'], alpha=0.3)

plt.title('Seaborn Area Chart (Age)')
plt.xlabel('Index')
plt.ylabel('Age')
plt.show()
```

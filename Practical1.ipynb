```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("facebook.csv", sep=';')
data
```

```python
subset1 = data[["Type", "comment", "like", "share"]]
subset2 = data[["Type", "Post Month", "Post Weekday", "Post Hour"]]
print(subset1)
print(subset2)
```

```python
subset3 = data.loc[0:5, ["Type", "comment", "like"]]
subset3
subset4 = data.iloc[0:5,[0,1]]
subset4
```

```python
merged_data = subset1.merge(subset2, on='Type', how='left')
print(merged_data)
```

```python
merged_data = pd.concat([subset1, subset2], axis=1)
print(merged_data)
```

```python
sorted_data = merged_data.sort_values(by="like", ascending=False, ignore_index=True)
sorted_data
```

```python
concat = pd.concat([subset3, subset4], axis=0)
concat
```

```python
transposed_data = sorted_data.transpose()
print(transposed_data);
```
